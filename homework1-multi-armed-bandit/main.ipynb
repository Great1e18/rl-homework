# 实验一：多臂老虎机——智慧仓库拣选口选择

## 一、场景
某智慧仓库有 5 个拣选口，每个拣选口处理订单的成功率固定但未知。
每次订单到来时，系统选择一个拣选口：
- 成功奖励 1
- 失败奖励 0
## 二、实验设置
臂数：K = 5
各拣选口真实成功率：probs = [0.3, 0.5, 0.7, 0.4, 0.6]
最优臂成功率：0.7
每次实验决策步数：T = 1000
独立重复实验次数：R = 200
奖励分布：伯努利分布。选择臂 a 后，以概率 probs[a] 获得奖励1，否则获得0。
遗憾定义：最优期望奖励 - 实际获得奖励，即 0.7 - reward
目标：在 1000 次决策中最大化累计成功次数。
