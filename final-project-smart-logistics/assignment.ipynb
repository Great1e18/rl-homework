一、作业题目
基于强化学习的智慧物流决策系统设计与实验

学生可从以下方向中任选一个（也可自拟，需教师审核）：

多 AGV 仓库路径规划与避障

多 AGV 任务调度与订单分配

快递分拨中心动态分拣策略

配送车辆路径优化

仓库库存补货决策

拣选口动态选择与人员调度

无人机配送路径与能耗优化

二、作业目标
综合运用本课程所学的多臂老虎机、马尔可夫过程、动态规划等知识，完成一个强化学习项目的完整流程：

问题建模 → 环境搭建 → 算法调用 → 仿真实验 → 结果分析 → 方案展示

本作业不要求你从零实现强化学习算法。你可以调用现成的算法库（如 Stable-Baselines3、CleanRL、RLlib 等），但必须清楚说明：

你如何把物流问题转化为 MDP；

你选择了什么算法、为什么选它；

你如何设计奖励函数；

你如何评价实验结果。

三、基本要求
1. 问题建模（必须）
用 MDP 五元组清晰描述你的问题：

要素	要求
状态 S	说明状态包含哪些信息，维度是多少
动作 A	说明动作空间是离散还是连续，每个动作的物理含义
奖励 R	给出奖励函数的具体表达式，并解释每一项的设计动机
转移 P	说明环境是确定性的还是随机的，转移规则是什么
折扣因子 γ	给出取值并说明理由
2. 环境搭建

环境可以是自定义的，也可以基于开源环境修改（需注明来源）。

至少实现：

reset()：重置环境，返回初始观测；

step(action)：执行动作，返回 (observation, reward, terminated, truncated, info)；

observation_space 和 action_space：正确定义

鼓励加入随机性（如订单随机到达、设备随机故障、路径随机拥堵）。

3. 算法调用（允许使用现成库）

必须说明：算法名称、关键超参数、训练步数、硬件环境。

不要求推导算法公式或手写算法核心代码。

4. 基线对比（必须）
至少设计一个基线策略，与强化学习策略对比。基线可以是：

随机策略；

贪心策略（如总是选择最近的货位）；

规则策略（如固定优先级调度）；

课程实验中的动态规划最优策略（若模型已知）。

对比指标建议包括：累计奖励、任务成功率、平均完成时间、碰撞次数、缺货率、能耗等。

5. 仿真实验（必须）
至少运行 5 个随机种子，报告平均值和标准差。

绘制以下图表（至少 3 张）：

学习曲线（训练奖励随步数变化）；

与基线的指标对比柱状图；

策略可视化（如 AGV 轨迹图、调度甘特图、热力图等）。

实验需可复现：固定随机种子，说明运行环境。

6. 成果交付
交付物	要求
代码	可运行，含 README.md 说明环境安装与运行方式
项目报告	8—12 页，结构见下文
答辩 PPT	8—10 分钟，重点讲建模思路、实验结果和反思
四、项目报告建议结构
问题背景与意义（1 页）
所选物流场景的现实痛点；
为什么适合用强化学习解决。

场景描述与 MDP 建模（2 页）

场景示意图；

状态、动作、奖励、转移、折扣因子的定义；

奖励函数设计动机。

环境实现（1—2 页）

关键代码片段（不超过 30 行）；

随机性设计。

算法选择与训练设置（1 页）

所选算法及理由；

超参数表；

训练环境与步数。

学习曲线；
